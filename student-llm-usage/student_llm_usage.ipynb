```python
import pandas as pd

df = pd.read_csv("/Labeled_prompts.csv.csv")
df.head()
```

```python
import matplotlib.pyplot as plt

counts = df["label"].value_counts()
print(counts)

plt.figure(figsize=(8, 5))
counts.plot(kind="bar", color="steelblue")
plt.title("What Do Students Ask AI Assistants For?")
plt.xlabel("Request type")
plt.ylabel("Number of prompts")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.savefig("chart_label_counts.png", dpi=200)
plt.show()
```

```python
table = pd.crosstab(df["subject"], df["label"])
print(table)

table.plot(kind="bar", stacked=True, figsize=(10, 6), colormap="tab10")
plt.title("Request Types by Subject")
plt.xlabel("Subject")
plt.ylabel("Number of prompts")
plt.xticks(rotation=30, ha="right")
plt.legend(title="Request type", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.savefig("chart_by_subject.png", dpi=200)
plt.show()
```
