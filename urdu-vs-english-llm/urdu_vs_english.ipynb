```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/scores.csv")
print(df[["score_en", "score_ur"]].describe())

avg = df.groupby("category")[["score_en", "score_ur"]].mean()
print(avg)

avg.plot(kind="bar", figsize=(10, 6), color=["#2b5c8f", "#e76f51"], edgecolor="black")
plt.title("Average Score by Category: English vs. Urdu (Gemini Flash)")
plt.xlabel("Question category")
plt.ylabel("Average score (0 to 2)")
plt.ylim(0, 2.6)
plt.xticks(rotation=20, ha="right")
plt.legend(["English", "Urdu"], title="Question language", loc="upper right")
plt.tight_layout()
plt.savefig("scores_by_category.png", dpi=150)
plt.show()
```
